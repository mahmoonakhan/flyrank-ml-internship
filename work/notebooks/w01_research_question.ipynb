## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** Which pages should the content team review first for refresh, expansion, protection, or monitoring?

**Who acts on it:** A human content reviewer or editor looks at the top-ranked pages and decides whether to update the article, rewrite the title/meta, expand thin sections, or simply monitor.

**Cost of a wrong recommendation:**
- *False positive:* Recommending a page that is not actually declining wastes reviewer time — the team looks at a healthy page instead of a troubled one.
- *False negative:* Missing a truly declining page means it keeps losing search traffic while the team focuses elsewhere.
- The cost is **reviewer capacity and missed opportunity**, not a broken production system. This makes precision@K the right metric: we want the top of the queue to be genuinely worth looking at.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

These three numbers show why a ranked review queue is worth building:

1. **Over half the inventory is declining** — the problem is widespread, so a manual review of every page is impossible.
2. **Thousands of pages are stale but still visible** — they have audience demand (impressions) but haven't been updated in 6+ months, making them obvious refresh candidates.
3. **Thousands more have good position but underperform on CTR** — they rank well but fail to win clicks, suggesting metadata or snippet issues.

Together, these numbers prove there is a large, actionable pool of pages where ML-based prioritization can help a human reviewer spend time where it matters most.

```python
import pandas as pd

# Load the starter dataset directly from the public repo (no local files needed)
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Real number 1: How many pages are declining?
declining = (df['trend_direction'] == 'down').sum()
total = len(df)
print(f"1. Declining pages: {declining:,} / {total:,} ({declining/total:.1%})")

# Real number 2: Stale visible pages (old but still getting impressions)
stale_visible = ((df['days_since_last_update'] >= 180) & (df['impressions_90d'] >= 500)).sum()
print(f"2. Stale visible pages (≥180 days old, ≥500 impressions): {stale_visible:,}")

# Real number 3: Low-CTR visible pages (high impressions, decent position, but poor CTR)
low_ctr = ((df['impressions_90d'] >= 500) &
           (df['avg_position'] > 0) &
           (df['avg_position'] <= 20) &
           (df['ctr'] < 0.5)).sum()
print(f"3. Low-CTR visible pages (≥500 imp, position 1-20, CTR < 0.5%): {low_ctr:,}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What I can claim:**
- *Observed / directional:* I can say which pages currently show signals associated with decline, staleness, or low CTR based on the data window.
- *Decision-support:* I can rank pages so a human reviewer spends time on the most promising candidates first.
- *Measured:* I can report precision@K and compare a model against a transparent baseline on held-out data.

**What I can NEVER claim:**
- *Causal proof:* I cannot say that refreshing a page *caused* a recovery. The data is observational, not experimental.
- *Predicting Google:* I cannot claim to have discovered a Google ranking factor or to be "predicting the algorithm."
- *Guaranteed outcome:* I cannot promise that acting on a recommendation will definitely improve traffic.
- *AI citations / rankings:* If I touch AI-referral data, I can only discuss click-through patterns, not whether an AI platform "cites" or "ranks" the content.

Every output will be framed as: *"Based on observed signals in this window, these pages are the strongest candidates for human review."*
